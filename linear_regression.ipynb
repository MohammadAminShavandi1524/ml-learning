```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("insurance.csv")

x = df["age"]
y = df["charges"]

# Mean
mean_x = x.mean()
mean_y = y.mean()

# Variance and Covariance
variance = x.var(ddof=0)
covariance = x.cov(y, ddof=0)

# Slope
m = covariance / variance

# Intercept
b = mean_y - m * mean_x

# Linear Regression
y_pred = m * x + b

print("Slope:", m)
print("Intercept:", b)

# Visualization
plt.figure(figsize=(12, 6))

plt.scatter(x, y, s=3, color="red")
plt.plot(x, y_pred, color="blue", linewidth=2)

plt.xlabel("Age")
plt.ylabel("Charges")
plt.title("Linear Regression")

plt.show()
```

```python
residuals = y - y_pred

absolute_errors = abs(residuals)

mae = sum(absolute_errors) / len(y)

print("MAE:", mae)
```
